# Results – FCBG Connectome Registration
Visualisation of metabolite concentration maps (MRSI) overlaid on the T1w UNI-DEN background image.

- **Background**: `sub-01_ses-01_acq-UNIDEN_T1w.nii` — the denoised uniform T1w image from MP2RAGE
- **Overlays**: individual metabolite maps from `mrs/`

We use **nilearn** (`plot_stat_map`) which resamples the overlay to the background on the fly, exactly like FSLeyes does.

```python
import os
import nibabel as nib
import numpy as np
import matplotlib.pyplot as plt
from nilearn import plotting, image

# ── Paths ────────────────────────────────────────────────────────────────────
PROJECT_ROOT = os.path.dirname(os.path.abspath("__file__"))
BIDS_DIR     = os.path.join(PROJECT_ROOT, "data", "bids")
SUBJ         = "sub-01"
SES          = "ses-01"

ANAT_DIR = os.path.join(BIDS_DIR, SUBJ, SES, "anat")
MRS_DIR  = os.path.join(BIDS_DIR, SUBJ, SES, "mrs")

# Background: UNI-DEN T1w
T1W_PATH = os.path.join(ANAT_DIR, f"{SUBJ}_{SES}_acq-UNIDEN_T1w.nii")

# List all metabolite maps
mrs_files = sorted(f for f in os.listdir(MRS_DIR) if f.endswith(".nii.gz"))
print(f"T1w background : {os.path.basename(T1W_PATH)}")
print(f"\nFound {len(mrs_files)} metabolite maps:")
for f in mrs_files:
    print(f"  {f}")
```

## 1 – Quick sanity check: image shapes & voxel sizes

Load the T1w and one MRS map, print their headers so we know if they live in the same space or need resampling.

```python
t1w_img = nib.load(T1W_PATH)
mrs_example = nib.load(os.path.join(MRS_DIR, mrs_files[0]))

def img_info(img, label):
    vox = np.sqrt(np.sum(img.affine[:3, :3] ** 2, axis=0))
    print(f"{label}")
    print(f"  shape      : {img.shape}")
    print(f"  voxel size : {vox.round(3)} mm")
    print(f"  dtype      : {img.get_data_dtype()}")
    print()

img_info(t1w_img,    "T1w UNI-DEN")
img_info(mrs_example, f"MRS – {mrs_files[0]}")
```

## 2 – View one metabolite map overlaid on the T1w

`plot_stat_map` from nilearn resamples the overlay to the background automatically and gives you the three standard views (axial / coronal / sagittal) with a colour-bar, just like FSLeyes.

```python
def metabolite_name(filename: str) -> str:
    """Extract the metabolite label from the BIDS filename."""
    # e.g. sub-01_ses-01_acq-OrigRes_desc-NAA_mrsi.nii.gz  →  NAA
    m = __import__("re").search(r"desc-([^_]+)_mrsi", filename)
    return m.group(1) if m else filename

def plot_overlay(mrs_filename: str, t1w_path: str = T1W_PATH,
                 cmap: str = "hot", threshold: float = None,
                 vmax: float = None):
    """
    Overlay a single metabolite map on the T1w background.
    threshold : lower bound for the colour scale (None = auto)
    vmax      : upper bound (None = 99th percentile of the map)
    """
    mrs_path = os.path.join(MRS_DIR, mrs_filename)
    mrs_img  = nib.load(mrs_path)
    label    = metabolite_name(mrs_filename)

    # Auto vmax at 99th percentile to avoid outlier bias
    if vmax is None:
        data = mrs_img.get_fdata()
        vmax = float(np.nanpercentile(data[data > 0], 99)) if np.any(data > 0) else None

    display = plotting.plot_stat_map(
        stat_map_img  = mrs_img,
        bg_img        = t1w_path,
        title         = f"{SUBJ} – {label}",
        display_mode  = "ortho",       # axial + coronal + sagittal
        cmap          = cmap,
        colorbar      = True,
        threshold     = threshold,
        vmax          = vmax,
        resampling_interpolation = "continuous",
    )
    plt.show()
    return display

# Show NAA as a first example
plot_overlay("sub-01_ses-01_acq-OrigRes_desc-NAA_mrsi.nii.gz")
```

## 3 – Coverage check: do the MRSI maps cover the whole brain?

Here we overlay **several metabolite maps** on the T1w in a single figure — one row per map — so you can directly compare spatial coverage across metabolites.

```python
# Pick a representative subset of metabolites for the coverage grid
COVERAGE_MAPS = [
    "sub-01_ses-01_acq-OrigRes_desc-NAA_mrsi.nii.gz",
    "sub-01_ses-01_acq-OrigRes_desc-Cr_mrsi.nii.gz",
    "sub-01_ses-01_acq-OrigRes_desc-Glu_mrsi.nii.gz",
    "sub-01_ses-01_acq-OrigRes_desc-Ins_mrsi.nii.gz",
    "sub-01_ses-01_acq-OrigRes_desc-GABA_mrsi.nii.gz",
    "sub-01_ses-01_acq-OrigRes_desc-GluaddGln_mrsi.nii.gz",
]

# We use a fixed axial slice near the centre of the brain
t1w_img  = nib.load(T1W_PATH)
t1w_data = t1w_img.get_fdata()
mid_z    = t1w_data.shape[2] // 2   # axial midpoint

fig, axes = plt.subplots(
    nrows=len(COVERAGE_MAPS), ncols=3,
    figsize=(14, 4 * len(COVERAGE_MAPS)),
)
if len(COVERAGE_MAPS) == 1:
    axes = [axes]   # keep iterable

SLICE_INDICES = {
    "axial":    ("z", mid_z),
    "coronal":  ("y", t1w_data.shape[1] // 2),
    "sagittal": ("x", t1w_data.shape[0] // 2),
}
view_keys = list(SLICE_INDICES.keys())

for row_idx, mrs_filename in enumerate(COVERAGE_MAPS):
    mrs_path  = os.path.join(MRS_DIR, mrs_filename)
    mrs_img   = nib.load(mrs_path)
    label     = metabolite_name(mrs_filename)

    data = mrs_img.get_fdata()
    vmax = float(np.nanpercentile(data[data > 0], 99)) if np.any(data > 0) else 1.0

    for col_idx, view in enumerate(view_keys):
        ax = axes[row_idx][col_idx]
        cut_coord_key, cut_val = SLICE_INDICES[view]

        # Convert voxel index to MNI/world coordinate
        coord = nib.affines.apply_affine(t1w_img.affine, [
            cut_val if cut_coord_key == "x" else t1w_data.shape[0] // 2,
            cut_val if cut_coord_key == "y" else t1w_data.shape[1] // 2,
            cut_val if cut_coord_key == "z" else t1w_data.shape[2] // 2,
        ])
        world_cut = float(coord[{"x": 0, "y": 1, "z": 2}[cut_coord_key]])

        display = plotting.plot_stat_map(
            stat_map_img  = mrs_img,
            bg_img        = T1W_PATH,
            display_mode  = cut_coord_key,
            cut_coords    = [world_cut],
            cmap          = "hot",
            colorbar      = True,
            vmax          = vmax,
            threshold     = 1e-6,
            axes          = ax,
            title         = f"{label} – {view}",
            resampling_interpolation = "continuous",
        )

fig.suptitle(f"{SUBJ} – MRSI coverage check (T1w UNI-DEN background)", fontsize=14, y=1.01)
plt.tight_layout()
plt.show()
```

## 4 – Interactive viewer: pick any metabolite

Change `METABOLITE` below to any label (e.g. `"Cr"`, `"Glu"`, `"GABA"`, `"Ins"`) and re-run the cell.

```python
METABOLITE = "NAA"   # ← change me

# Find the matching file
match = [f for f in mrs_files if f"desc-{METABOLITE}_mrsi" in f]
if not match:
    available = [metabolite_name(f) for f in mrs_files]
    raise ValueError(f"Metabolite '{METABOLITE}' not found. Available: {available}")

print(f"Showing: {match[0]}")
plot_overlay(match[0], cmap="hot")
```
